# Project: Time Series Forecasting in Retail (M5 Competition Dataset)

## Objective

This project requires students to select **one department** from **one store** in one of the states available in the [M5 Competition dataset](https://www.kaggle.com/competitions/m5-forecasting-accuracy) (Walmart hierarchical sales data), which was studied in class.

The M5 dataset contains daily unit sales for 3,049 items across three main categories (Hobbies, Foods, Household), further divided into seven departments: Foods1, Foods2, Foods3, Hobbies1, Hobbies2, Household1, Household2. Items are sold in 10 stores across three states: California (CA1–CA4), Texas (TX1–TX3), and Wisconsin (WI1–WI3). The time series span 1,969 days (January 29, 2011 to June 19, 2016).

The overall goal is to apply and compare a wide range of time series forecasting methods covered in the course — from simple baselines to statistical, machine learning, deep learning, foundation models, and automated frameworks (AutoGluon) — while producing accurate point forecasts and probabilistic forecasts for a **28-day horizon**, matching the M5 Competition setup.

The project is divided into **three sequential phases**, each submitted as a separate Jupyter Notebook. Each phase builds on the previous one (same selected department–store series, same EDA foundation).

## Phase Structure and Deliverables

### Phase 1 – Statistical Models  
**Focus:** Classical and statistical time series methods  
**Deliverable:** Jupyter Notebook named `M5_Forecasting_Phase1_StatisticalModels_[groupname].ipynb`

**Tasks:**
- Select one department–store combination (e.g., Foods3_CA1, Hobbies1_TX2, etc.) and clearly state your choice on the first page.
- Download and preprocess the relevant time series data (cleaning, handling missing values, outliers/anomalies, calendar/event alignment, etc.).
- Perform comprehensive **Exploratory Data Analysis (EDA)**: visualize trends, seasonality (weekly, yearly), autocorrelation, decomposition, stationarity tests, anomaly detection, impact of holidays/SNAP events/prices, etc.
- Apply and compare **all statistical/baseline methods** covered in class, including:
  - Naïve
  - Seasonal Naïve
  - Historic Average
  - Random Walk With Drift
  - ARIMA (with exogenous regressors)
  - ETS
- Evaluate models using appropriate metrics (e.g., MAE, RMSE, MAPE, MASE, RMSSE, etc).
- Compare performance on a validation set (last 28 days of training period).
- Present and visualize point forecasts and probabilistic forecasts (if possible) from the best statistical model with visualizations.
- Test the impact of including exogenous regressors (prices, SNAP, calendar events) vs. univariate version.

**Assessment for Phase 1 (out of 100 points – scaled for final grade):**
- EDA quality & visualizations – 25 points
- Correct implementation of statistical models – 35 points
- Evaluation metrics and model comparison – 25 points
- Documentation, code quality, clarity – 15 points
