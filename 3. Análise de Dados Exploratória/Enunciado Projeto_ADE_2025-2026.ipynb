# Projeto de Análise de Dados Exploratória: Segmentação de Clientes e Sistema de Recomendação para um Retalhista

## Descrição do problema:

Este projeto utiliza um conjunto de dados relativo a vendas de um retalhista. O principal objetivo é tornar as estratégias de *marketing* mais eficientes e aumentar as vendas através da segmentação dos clientes. Pretende-se transformar os dados das vendas num conjunto de informação focada no cliente, criando novas *features* que facilitarão a segmentação dos clientes em grupos distintos. Esta segmentação permitirá compreender os perfis e as preferências dos diferentes grupos de clientes. Com base nesta segmentação dos clientes, pretende-se implementar um "sistema de recomendação" que irá sugerir os produtos mais vendidos aos clientes de cada segmento que ainda não compraram esses artigos, melhorando assim a eficácia do *marketing* e promovendo o aumento das vendas.

## Descrição do conjunto de dados:

O conjunto de dados inclui vários campos relacionados com as vendas do retalhista. As *features* do conjunto de dados são descritas na tabela seguinte:

| __*Feature*__ | __Descrição__ |
| :--- | :--- |      
| __N_Fatura__ | Código único que identifica a venda. Se este código começar com a letra "c", indica uma anulação. |
| __SKU__ | Código único do produto (SKU - Stock Keeping Unit). |
| __Descricao__  | Descrição do produto. |
| __Quantidade__ | Número de unidades vendidas do produto na venda. |
| __Data_Fatura__ | Data e a hora da venda. |
| __Preco_Unitario__ | Preço unitário do produto (em euros). |
| __Cliente__ | Identificador único atribuído a cada cliente. |
| __Pais__ | País do cliente. |

## Objetivos do Projeto:

Os objetivos do projeto incluem os seguintes tópicos:

1. **Limpeza e transformação de dados:** Limpar o conjunto de dados através do tratamento de valores em falta, duplicados e *outliers*, preparando-o de forma eficaz para as fases seguintes do projeto.

1. **Engenharia de *features*:** Desenvolver novas *features* com base nos dados das vendas para criar um **conjunto de dados centrado no cliente**, nomeadamente,

  + número de dias desde a última compra,
  + intervalo de tempo médio entre compras (em dias),
  + dia da semana preferido para compra,
  + hora do dia preferido para a compra,
  + número total de compras,
  + quantia total gasta,
  + número total de produtos distintos comprados,
  + gasto médio por compra,
  + número médio de produtos por compra,
  + gasto médio mensal,
  + desvio padrão do gasto mensal,
  + representação numérica da tendência de compra ao longo do período histórico.

  estabelecendo as bases para a segmentação de clientes.

1. **Pré-processamento de dados:** Efetuar o *scaling* das *features* e a redução da dimensão para simplificar os dados, aumentando a eficiência do processo de *clustering*.

1. **Segmentação de clientes:** Segmentar os clientes em grupos distintos utilizando o algoritmo de *clustering* K-Means, facilitando o *marketing* direcionado e as estratégias personalizadas.

1. **Análise e avaliação de *clusters*:** Analisar e traçar o perfil de cada *cluster* para desenvolver estratégias de *marketing* direcionadas e avaliar a qualidade dos *clusters* formados.

1. **Sistema de Recomendação:** Implementar um sistema para recomendar os produtos mais vendidos do *cluster* aos clientes desse *cluster* que ainda não compraram esses produtos, com o objetivo de aumentar a eficácia das vendas e do *marketing*. Opcional: comparar esta implementação com um sistema de recomendação desenvolvido utilizando as técnicas apresentadas na UC.
